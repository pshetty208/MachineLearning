```python
import numpy as np
import math
Input_layer = [float(item) for item in input("Enter the input neurons (seprated by ',') : ").split(',')]
n_hidden_layers = int(input("Enter the number of hidden layer: "))
Func_type = input("Enter the activation function (s for sigmoid, r for relu, lr for leaky relu) : ")
learning_rate = float(input("Enter the learning rate : "))
```

```python
def activation_func(input,type):
    if(type == "s"):
        out = (1/(1+np.exp(-input)))
    elif(type == "r"):
        out = max(0,input)
    elif(type == "lr"):
        out = max(0.01,input)
    return out
```

```python
def der_activation_func(input,type):
    if(type == "s"):
        out = input*(1-input)
        print("\tder 2 =", str(input), "*(1 -", str(input),") = ", out)
    elif(type == "r"):
        if(input>0) :
            out = 1
            print("\tder 2 ->", str(input),"is > 0 , therefore =", out)
        else :
            out = 0
            print("\tder 2 -> input is <= 0 , therefore = 0")
    elif(type == "lr"):
        if(input>0) :
            out = 1*0.01
            print("\tder 2 ->", str(input),"is > 0 , therefore =", str(input), "*0.01=", out)
        else :
            out = 0
            print("\tder 2 -> input is <= 0 , therefore = 0")
    return out
```

```python
#making 3 dictionaries to store the weights, netoutput(without actication) and output(with actication function)
weights = dict()
weights_netoutput = dict()
weights_output = dict()
```

```python
y = Input_layer.copy()
for every_layer in range(n_hidden_layers):
    y_new = []
    print("Hidden Layer :", every_layer+1)
    n_layer = int(input("Enter the number of neuron in this hidden layer : "))
    for neuron in range(n_layer) :
        g = 0
        string = ""
        print("\tNeuron :", neuron+1)
        bias = float(input("\tEnter the Bias for neuron : "))
        for n in range(len(y)) :
            print("\t\tFor weight from neuron", n+1)
            weight_name = input("\t\tEnter the weight name: ")
            weight = float(input("\t\tEnter the weight : "))
            weights[weight_name] = weight
            weights_netoutput[weight_name] = 'na'
            weights_output[weight_name] = 'na'
            g = g +  (weight*y[n])
            string = string + " + " + str((weight*y[n]))
        g = g + bias
        string = string + " + " + str(bias)
        ac = activation_func(g,Func_type)
        for key in weights_netoutput.keys():
            if weights_netoutput[key] == 'na':
                weights_netoutput[key] = g
        for key in weights_output.keys():
            if weights_output[key] == 'na':
                weights_output[key] = ac
        print("\tActivation function of", string, "(Net H",neuron+1,"_",every_layer+1,") =" , g, "is : (Out H",neuron+1,")", ac)
        y_new.append(ac)
        print("*****************************************************")
    y = y_new.copy() 
    print(y_new)
    print("===========================================================================")
```
