```python
import pandas as pd
train_data = pd.read_csv("traindata.csv", header = 0)
test_data = pd.read_csv("testdata.csv", header = 0)

print(train_data)
print(test_data)
```

```python
import numpy as np
# drop duplicated rows
train_data = train_data.drop_duplicates()
#print(train_data)
    
# replace zeros entries with np.nan
train_data = train_data.replace(0,np.nan)
train_data = train_data.dropna() # drop all nan entiries 
#print(train_data)

newCols = pd.get_dummies(train_data.iloc[:,0]) 
train_data = pd.concat([newCols, train_data], axis=1) # add them to the transfomed columns to the beggining  of the data frame
train_data = train_data.drop(train_data.columns[3], axis = 1) # drop nominal column (color)   

# drop duplicated rows
test_data= test_data.drop_duplicates()

# transform the nominal feature (color) to dummy
newCols=pd.get_dummies(test_data.iloc[:,0]) 
test_data = pd.concat([newCols, test_data], axis=1) # add them to the transfomed columns to the beggining  of the data frame
test_data= test_data.drop(test_data.columns[3], axis = 1) # drop nominal column (color)  

print("Test Data is", test_data)
print("train data is", train_data)
```

```python
Atrributes_train = list(train_data.columns)
print(Atrributes_train)
classes = train_data[Atrributes_train[-1]]
print(classes)
classes_name = classes.unique()
print(classes_name)

Atrributes_test = list(test_data.columns)
print(Atrributes_test)
```

```python
from sklearn import preprocessing

#train data
label_encoder = preprocessing.LabelEncoder()
for attribute in Atrributes_train[:-1] :
    if(train_data[attribute].dtypes == 'object') :
        train_data[attribute]= label_encoder.fit_transform(train_data[attribute]) 
    else :
        normalized_df=(train_data[attribute]-train_data[attribute].min())/(train_data[attribute].max()-train_data[attribute].min())
        train_data[attribute]= normalized_df
    
X_train = train_data[Atrributes_train[:-1]]
#X_train = preprocessing.normalize(X_train)
#X_train['Radius (cm)'] = preprocessing.normalize(X_train['Radius (cm)'])
Y_train = train_data[Atrributes_train[-1]]
print(X_train)
print(Y_train)

#test data
print("test_data")
label_encoder = preprocessing.LabelEncoder()
for attribute in Atrributes_test[:-2] :
    if(test_data[attribute].dtypes == 'object') :
        test_data[attribute]= label_encoder.fit_transform(test_data[attribute])
    else :
        normalized_df=(test_data[attribute]-test_data[attribute].min())/(test_data[attribute].max()-test_data[attribute].min())
        test_data[attribute]= normalized_df
    
X_test = test_data[Atrributes_test[:-2]]
Y_test = test_data[Atrributes_test[-1]]
print(X_test)
print(Y_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier

for n in range(1,12) :
    print(n)
    knn = KNeighborsClassifier(n_neighbors=n)
    knn.fit(X_train, Y_train)
    y_pred = knn.predict(X_test)
    print(y_pred)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(Y_test,y_pred)
print(cm)
```

```python
from sklearn.metrics import classification_report
print(classification_report(Y_test, y_pred, zero_division=0))
```
